### VectorAssembles
con el vector assembler basicamente le pasamos el inputCols con el nombre de las columnas que queremos unir y en el outputcol crea ese vector con las columnas unidas y le especificamos un nombre para esa columna. 
Luego simplemente seria guardar el resultado como otro data frame usando el ensamblador.transform y el dataframe que queremos transformar

```python
from pyspark.ml.feature import VectorAssembler
ensamblador = VectorAssembler(
    inputCols= ["Precio_dia_Anterior", "Retorno_diario"],
    outputCol="features"
)
df_procesado = ensamblador.transform(df_ml)
display(df_procesado)
```

# 1. Dividir los datos (Train / Test Split)

```python
from pyspark.sql  import SparkSession
#randomSplit sobre el dataframe 
train_data, test_data = df_procesado.randomSplit([0.8,0.2],seed=42)
#verificación de registro
print("Registro de entrenamiento: ",train_data.count())
print("Registro de prueba: ",test_data.count())
```

# Instanciar el algoritmo

```python
from pyspark.ml.regression import RandomForestRegressor
rf = RandomForestRegressor(featuresCol="features",labelCol="Close")
```

# Entrenar y predecir

```python
model = rf.fit(train_data)
prediccion = model.transform(test_data)
display(prediccion)
```
